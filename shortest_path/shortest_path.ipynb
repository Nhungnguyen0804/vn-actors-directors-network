```python
import json
import networkx as nx
from networkx.readwrite import json_graph

# 1. Load file đồ thị nhóm 

json_file = "../vn_film_collaboration_graph.json"  
json_file = "../vn_bipartitr_graph.json" 

with open(json_file, "r", encoding="utf-8") as f:
    data = json.load(f)

# Chuyển JSON thành Graph NetworkX
G = json_graph.node_link_graph(data)

print("Số lượng node:", G.number_of_nodes())
print("Số lượng cạnh:", G.number_of_edges())



# 2. Chọn 2 node để tìm đường đi



start_node = "Diệu Nhi"
end_node = "Hồng Nhung"

# 3. Tìm đường đi ngắn nhất


try:
    path = nx.shortest_path(G, source=start_node, target=end_node)
    print("\nĐường đi ngắn nhất giữa hai node:")
    print(" → ".join(path))

except nx.NetworkXNoPath:
    print(f"Không có đường đi giữa {start_node} và {end_node}")

except nx.NodeNotFound as e:
    print("Lỗi: Node không tồn tại trong đồ thị.")
    print(e)


import warnings
warnings.filterwarnings("ignore", category=FutureWarning)
```
